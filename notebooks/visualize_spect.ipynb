# Spectral Data Visualization

Simple notebook to visualize original .spect data files from Vera0925 dataset.

```python
import numpy as np
import matplotlib.pyplot as plt
import os

# Data file path
filename = "1360-1800-4_smoothed.txt"
data_dir = "../data/Vera0925/preprocessed"
file_path = os.path.join(data_dir, filename)
```

```python
# Load data file
data = np.loadtxt(file_path)

# First column is intensity, third column is wavelength
intensity = data[:, 1]
wavelength = data[:, 0]

print(f"File name: {filename}")
print(f"Number of data points: {len(wavelength)}")
print(f"Wavelength range: {wavelength.min():.1f} - {wavelength.max():.1f}")
print(f"Intensity range: {intensity.min():.6e} - {intensity.max():.6e}")
```

```python
# Visualize spectral data
fig = plt.figure(figsize=(12, 6))
plt.plot(wavelength, intensity, 'b-', linewidth=1.5)
plt.xlabel('Wavelength', fontsize=12)
plt.ylabel('Intensity', fontsize=12)
plt.title(f'Spectral Data - {filename}', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Output directory for saving plots (inside Vera0925 folder)
plot_dir = os.path.join(data_dir, "plot")
os.makedirs(plot_dir, exist_ok=True)  # Create directory if it doesn't exist

# Save the plot
base_filename = os.path.splitext(filename)[0]  # Remove .spect extension
save_path = os.path.join(plot_dir, f"{base_filename}.png")
fig.savefig(save_path, dpi=300, bbox_inches='tight')
print(f"Plot saved to: {save_path}")
```
