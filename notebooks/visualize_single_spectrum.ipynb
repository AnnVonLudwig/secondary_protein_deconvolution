# 单个光谱数据快速可视化

设置想查看的 `.spect` 文件名并依次运行下面的代码单元，即可读取并绘制原始光谱强度随波长的变化曲线。

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
```

```python
# === 配置：填写待查看的光谱文件名 ===
DATA_DIR = Path('../data/Vera0925')  # 数据文件夹路径
FILENAME = '2665-2995 on collagen.spect'  # 修改为想要查看的文件名

file_path = DATA_DIR / FILENAME
print(f'选定文件: {file_path}')
if not file_path.exists():
    raise FileNotFoundError(f'未找到文件: {file_path}')
```

```python
# 读取光谱数据
data = np.loadtxt(file_path)
wavelength = data[:, 2]  # 第三列：波长
intensity = data[:, 0]  # 第一列：强度

print(f'数据点数量: {len(wavelength)}')
print(f'波长范围: {wavelength.min():.2f} - {wavelength.max():.2f}')
print(f'强度范围: {intensity.min():.6f} - {intensity.max():.6f}')
```

```python
# 绘制光谱曲线
plt.figure(figsize=(12, 6))
plt.plot(wavelength, intensity, color='C0', linewidth=1.5)
plt.xlabel('Wavelength')
plt.ylabel('Intensity')
plt.title(f'Spectral Data: {FILENAME}')
plt.tight_layout()
plt.show()
```
